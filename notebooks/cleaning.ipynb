# Data Cleaning & ETL Pipeline

This notebook focuses on cleaning the raw e-commerce dataset and preparing it for analysis.

Steps performed:
- Handling missing values
- Removing duplicates
- Filtering invalid data
- Feature engineering

### Importing Required Libraries and Loading Dataset

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/online_retail_II.csv")
df.head()
```

# Basic Info

```python
df.info()
df.describe()
df.isnull().sum()
```

# Data Cleaning

```python
df = df.dropna(subset=['Customer ID'])

df = df.drop_duplicates()

df = df[df['Quantity'] > 0]

df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['Customer ID'] = df['Customer ID'].astype(int)

df = df[df['Price'] > 0]
```

### Total Price

Calculating total revenue per transaction.

```python
df['TotalPrice'] = df['Quantity'] * df['Price']
```

### Date Features

Extracting year, month, and day for time-based analysis.

```python
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['Day'] = df['InvoiceDate'].dt.day
```

## Customer-Level Analysis

Aggregating data to understand customer purchasing behavior.

```python
customer_df = df.groupby('Customer ID').agg({
    'TotalPrice': 'sum',
    'Invoice': 'count'
}).rename(columns={
    'TotalPrice': 'TotalSpent',
    'Invoice': 'TotalOrders'
}).reset_index()

customer_df.head()
```

## Final Data Validation

Ensuring cleaned dataset is consistent and free from major issues.

```python
df.shape
```

```python
df.isnull().sum()
```

## Saving Cleaned Data

Exporting cleaned datasets for further analysis and visualization.

```python
df.to_csv("../data/processed/cleaned_data.csv", index=False)
customer_df.to_csv("../data/processed/customer_data.csv", index=False)
```

The dataset has been cleaned and transformed successfully. It is now ready for exploratory data analysis and visualization.
